```python
# ==============================================================================
# DASHBOARD INTERACTIVO DE OPTIMIZACIÓN DE CONSTELACIÓN DE SATÉLITES EN 3D
# ==============================================================================

from IPython.display import HTML, display

html_code = """
<!DOCTYPE html>
<html lang="es">
<head>
  <meta charset="UTF-8">
  <title>AstroShield - Dashboard 3D</title>
  <script src="https://cdnjs.cloudflare.com/ajax/libs/three.js/r128/three.min.js"></script>
  <script src="https://cdn.jsdelivr.net/npm/three@0.128.0/examples/js/controls/OrbitControls.js"></script>
  <script src="https://cdn.jsdelivr.net/npm/chart.js"></script>
  <link href="https://fonts.googleapis.com/css2?family=Orbitron:wght@400;600;800&family=Rajdhani:wght@400;500;700&display=swap" rel="stylesheet">

  <style>
    /* Forzar pantalla completa absoluta dentro de Colab */
    html, body {
      margin: 0 !important;
      padding: 0 !important;
      width: 100vw !important;
      height: 100vh !important;
      overflow: hidden !important;
      background-color: #030712 !important;
      font-family: 'Rajdhani', sans-serif;
      color: #e2e8f0;
    }
    #canvas-container {
      width: 100vw;
      height: 100vh;
      position: fixed;
      top: 0;
      left: 0;
      z-index: 1;
    }

    .hud-layer {
      position: fixed;
      top: 0;
      left: 0;
      width: 100vw;
      height: 100vh;
      z-index: 10;
      pointer-events: none;
      display: grid;
      grid-template-columns: 320px 1fr 340px;
      grid-template-rows: 60px 1fr 200px;
      gap: 15px;
      padding: 15px;
      box-sizing: border-box;
    }

    .interactive { pointer-events: auto; }

    .panel {
      background: rgba(10, 16, 30, 0.85);
      border: 1px solid rgba(0, 229, 255, 0.25);
      border-radius: 8px;
      backdrop-filter: blur(12px);
      box-shadow: 0 8px 32px 0 rgba(0, 0, 0, 0.6);
      padding: 15px;
      display: flex;
      flex-direction: column;
      gap: 10px;
      overflow-y: auto;
    }

    header {
      grid-column: 1 / -1;
      display: flex;
      justify-content: space-between;
      align-items: center;
      background: rgba(10,16,30,0.95);
      border-bottom: 2px solid #00e5ff;
      padding: 0 20px;
      border-radius: 6px;
    }

    .brand { display: flex; align-items: center; gap: 12px; }
    .brand-icon {
      width: 28px; height: 28px; background: #00e5ff;
      clip-path: polygon(50% 0%, 100% 25%, 100% 75%, 50% 100%, 0% 75%, 0% 25%);
      box-shadow: 0 0 12px #00e5ff;
    }
    .brand-title {
      font-family: 'Orbitron', sans-serif;
      font-size: 1.1rem; font-weight: 800; color: #fff;
      letter-spacing: 2px;
    }

    .status-badge {
      font-family: 'Orbitron', sans-serif; font-size: 0.7rem;
      padding: 4px 10px; border-radius: 20px;
      background: rgba(16, 185, 129, 0.15);
      border: 1px solid #10b981; color: #34d399;
      display: flex; align-items: center; gap: 6px;
    }
    .status-dot {
      width: 8px; height: 8px; background: #10b981; border-radius: 50%;
      box-shadow: 0 0 8px #10b981;
    }

    .section-title {
      font-family: 'Orbitron', sans-serif; font-size: 0.8rem;
      color: #00e5ff; text-transform: uppercase; letter-spacing: 1px;
      border-bottom: 1px solid rgba(0, 229, 255, 0.2); padding-bottom: 4px;
    }

    .left-panel { grid-row: 2 / 4; }
    .right-panel { grid-row: 2 / 4; }
    .bottom-panel { grid-column: 2 / 3; grid-row: 3 / 4; }

    .metric-card {
      background: rgba(15, 23, 42, 0.6);
      border: 1px solid rgba(255, 255, 255, 0.05);
      padding: 8px 10px; border-radius: 6px;
      display: flex; flex-direction: column;
    }
    .metric-label { font-size: 0.7rem; color: #94a3b8; text-transform: uppercase; }
    .metric-value { font-family: 'Orbitron', sans-serif; font-size: 1.2rem; color: #fff; font-weight: 700; }

    .btn {
      background: linear-gradient(135deg, #00e5ff 0%, #0077ff 100%);
      color: #000; font-family: 'Orbitron', sans-serif; font-weight: 700; font-size: 0.75rem;
      border: none; padding: 8px 12px; border-radius: 4px; cursor: pointer;
      text-transform: uppercase; letter-spacing: 1px;
    }
    .btn-secondary {
      background: rgba(30, 41, 59, 0.8); color: #00e5ff;
      border: 1px solid #00e5ff;
    }

    .list-item {
      display: flex; justify-content: space-between; align-items: center;
      background: rgba(15, 23, 42, 0.5); padding: 6px 8px; border-radius: 4px; font-size: 0.75rem;
      border-left: 3px solid #3b82f6;
    }
    .list-item.covered { border-left-color: #10b981; }
    .list-item.danger { border-left-color: #ef4444; }

    .badge { font-size: 0.6rem; padding: 2px 5px; border-radius: 3px; font-weight: bold; }
    .badge-high { background: rgba(239, 68, 68, 0.2); color: #f87171; border: 1px solid #ef4444; }
    .badge-med { background: rgba(245, 158, 11, 0.2); color: #fbbf24; border: 1px solid #f59e0b; }

    .chart-container { position: relative; flex: 1; width: 100%; min-height: 90px; }
  </style>
</head>
<body>

  <div id="canvas-container"></div>

  <div class="hud-layer">
    <header class="interactive">
      <div class="brand">
        <div class="brand-icon"></div>
        <div class="brand-title">ASTROSHIELD AI // CONTROL DE MISIÓN</div>
      </div>
      <div class="status-badge">
        <div class="status-dot"></div>
        <span>SISTEMA ACTIVO</span>
      </div>
    </header>

    <div class="panel left-panel interactive">
      <div class="section-title">Parámetros</div>
      <div class="metric-card">
        <span class="metric-label">Constelación</span>
        <span class="metric-value">8 SATÉLITES</span>
      </div>
      <div class="metric-card">
        <span class="metric-label">Generación</span>
        <span class="metric-value" id="gen-counter">0 / 100</span>
      </div>
      <div class="metric-card">
        <span class="metric-label">Fitness Óptimo</span>
        <span class="metric-value" id="fitness-value" style="color: #00e5ff;">0.00</span>
      </div>
      <div class="metric-card">
        <span class="metric-label">Cobertura</span>
        <span class="metric-value" id="coverage-value" style="color: #10b981;">0 %</span>
      </div>

      <button class="btn" id="btn-run-ga">▶ Ejecutar Evolución</button>
      <button class="btn btn-secondary" id="btn-reset">↺ Reiniciar</button>
      <button class="btn btn-secondary" id="btn-toggle-orbit">⏸ Pausar Órbitas</button>
    </div>

    <div class="panel right-panel interactive">
      <div class="section-title">Regiones Objetivo (6)</div>
      <div id="regions-list" style="display:flex; flex-direction:column; gap:5px;"></div>
      <div class="section-title" style="margin-top: 10px;">Zonas Restringidas (3)</div>
      <div id="debris-list" style="display:flex; flex-direction:column; gap:5px;"></div>
    </div>

    <div class="panel bottom-panel interactive">
      <div class="section-title">Convergencia del Fitness</div>
      <div class="chart-container">
        <canvas id="fitnessChart"></canvas>
      </div>
    </div>
  </div>

  <script>
    const EARTH_RADIUS = 5;
    const SAT_COUNT = 8;

    const REGIONS = [
      { name: "Península de Yucatán", lat: 20.5, lon: -89.0, priority: "Alta", weight: 3.0 },
      { name: "Amazonía Brasileña", lat: -3.5, lon: -62.0, priority: "Alta", weight: 3.0 },
      { name: "California, EE.UU.", lat: 36.8, lon: -119.4, priority: "Media", weight: 2.0 },
      { name: "Región del Sahel", lat: 15.0, lon: 0.0, priority: "Media", weight: 2.0 },
      { name: "Japón", lat: 36.0, lon: 138.0, priority: "Media", weight: 2.0 },
      { name: "Interior de Australia", lat: -25.0, lon: 133.0, priority: "Normal", weight: 1.0 }
    ];

    const DEBRIS_ZONES = [
      { name: "Atlántico Norte", lat: 45.0, lon: -35.0, radiusDeg: 12.0 },
      { name: "Pacífico Sur", lat: -30.0, lon: -120.0, radiusDeg: 10.0 },
      { name: "Región Polar Norte", lat: 78.0, lon: 30.0, radiusDeg: 8.0 }
    ];

    function latLonToVector3(lat, lon, radius) {
      const phi = (90 - lat) * (Math.PI / 180);
      const theta = (lon + 180) * (Math.PI / 180);
      const x = -(radius * Math.sin(phi) * Math.cos(theta));
      const z = (radius * Math.sin(phi) * Math.sin(theta));
      const y = (radius * Math.cos(phi));
      return new THREE.Vector3(x, y, z);
    }

    function angularDistanceDeg(lat1, lon1, lat2, lon2) {
      const rad1 = lat1 * Math.PI / 180;
      const rad2 = lat2 * Math.PI / 180;
      const dLon = (lon2 - lon1) * Math.PI / 180;
      const val = Math.sin(rad1) * Math.sin(rad2) + Math.cos(rad1) * Math.cos(rad2) * Math.cos(dLon);
      return Math.acos(Math.max(-1, Math.min(1, val))) * (180 / Math.PI);
    }

    const container = document.getElementById('canvas-container');
    const scene = new THREE.Scene();
    scene.fog = new THREE.FogExp2(0x030712, 0.015);

    const camera = new THREE.PerspectiveCamera(45, window.innerWidth / window.innerHeight, 0.1, 1000);
    camera.position.set(0, 8, 16);

    const renderer = new THREE.WebGLRenderer({ antialias: true, alpha: true });
    renderer.setSize(window.innerWidth, window.innerHeight);
    renderer.setPixelRatio(window.devicePixelRatio);
    container.appendChild(renderer.domElement);

    const controls = new THREE.OrbitControls(camera, renderer.domElement);
    controls.enableDamping = true;
    controls.dampingFactor = 0.05;

    scene.add(new THREE.AmbientLight(0x334155, 1.5));
    const sunLight = new THREE.DirectionalLight(0xffffff, 2.0);
    sunLight.position.set(20, 10, 20);
    scene.add(sunLight);

    const earthGroup = new THREE.Group();
    scene.add(earthGroup);

    const textureLoader = new THREE.TextureLoader();
    const earthMesh = new THREE.Mesh(
      new THREE.SphereGeometry(EARTH_RADIUS, 64, 64),
      new THREE.MeshPhongMaterial({
        map: textureLoader.load('https://raw.githubusercontent.com/mrdoob/three.js/master/examples/textures/planets/earth_atmos_2048.jpg'),
        bumpMap: textureLoader.load('https://raw.githubusercontent.com/mrdoob/three.js/master/examples/textures/planets/earth_normal_2048.jpg'),
        specularMap: textureLoader.load('https://raw.githubusercontent.com/mrdoob/three.js/master/examples/textures/planets/earth_specular_2048.jpg'),
        specular: new THREE.Color(0x333333), shininess: 15
      })
    );
    earthGroup.add(earthMesh);

    DEBRIS_ZONES.forEach(zone => {
      const pos = latLonToVector3(zone.lat, zone.lon, EARTH_RADIUS + 0.02);
      const mesh = new THREE.Mesh(
        new THREE.SphereGeometry(EARTH_RADIUS * Math.sin(zone.radiusDeg * Math.PI / 180), 32, 32),
        new THREE.MeshBasicMaterial({ color: 0xef4444, transparent: true, opacity: 0.35, wireframe: true })
      );
      mesh.position.copy(pos);
      earthGroup.add(mesh);
    });

    const regionMeshes = [];
    REGIONS.forEach((reg, i) => {
      const pos = latLonToVector3(reg.lat, reg.lon, EARTH_RADIUS + 0.03);
      const mesh = new THREE.Mesh(new THREE.SphereGeometry(0.12, 16, 16), new THREE.MeshBasicMaterial({ color: 0xfbbf24 }));
      mesh.position.copy(pos);
      earthGroup.add(mesh);
      regionMeshes.push(mesh);
    });

    const satGroup = new THREE.Group();
    earthGroup.add(satGroup);
    const satMeshes = [];
    const orbitLines = [];

    for(let i=0; i<SAT_COUNT; i++) {
      const satContainer = new THREE.Group();
      satContainer.add(new THREE.Mesh(new THREE.OctahedronGeometry(0.12), new THREE.MeshBasicMaterial({ color: 0x00e5ff })));
      satContainer.add(new THREE.Mesh(new THREE.BoxGeometry(0.5, 0.02, 0.15), new THREE.MeshBasicMaterial({ color: 0x3b82f6 })));
      satGroup.add(satContainer);
      satMeshes.push(satContainer);

      const points = [];
      for(let j=0; j<=64; j++) {
        const angle = (j / 64) * Math.PI * 2;
        points.push(new THREE.Vector3(Math.cos(angle)*(EARTH_RADIUS+1.2), 0, Math.sin(angle)*(EARTH_RADIUS+1.2)));
      }
      const orbitLine = new THREE.Line(new THREE.BufferGeometry().setFromPoints(points), new THREE.LineBasicMaterial({ color: 0x00e5ff, transparent: true, opacity: 0.25 }));
      earthGroup.add(orbitLine);
      orbitLines.push(orbitLine);
    }

    let currentSats = Array.from({length: SAT_COUNT}, () => ({
      lat: (Math.random() - 0.5) * 140, lon: (Math.random() - 0.5) * 360, alt: 1.2
    }));

    function evaluateFitness(genome) {
      let score = 0, coveredCount = 0;
      REGIONS.forEach(reg => {
        let isCovered = false;
        genome.forEach(sat => {
          if(angularDistanceDeg(sat.lat, sat.lon, reg.lat, reg.lon) <= 12.0) isCovered = true;
        });
        if(isCovered) { score += (100 * reg.weight); coveredCount++; }
      });
      return { score, coveragePct: Math.round((coveredCount / REGIONS.length) * 100) };
    }

    function updateVisuals(genome) {
      genome.forEach((sat, i) => {
        const pos = latLonToVector3(sat.lat, sat.lon, EARTH_RADIUS + sat.alt);
        satMeshes[i].position.copy(pos);
        orbitLines[i].quaternion.setFromUnitVectors(new THREE.Vector3(0, 1, 0), pos.clone().normalize());
      });

      const evalRes = evaluateFitness(genome);
      REGIONS.forEach((reg, i) => {
        let isCovered = genome.some(sat => angularDistanceDeg(sat.lat, sat.lon, reg.lat, reg.lon) <= 12.0);
        regionMeshes[i].material.color.setHex(isCovered ? 0x10b981 : 0xef4444);
      });

      document.getElementById('fitness-value').innerText = evalRes.score.toFixed(1);
      document.getElementById('coverage-value').innerText = evalRes.coveragePct + " %";

      const regList = document.getElementById('regions-list');
      regList.innerHTML = '';
      REGIONS.forEach(reg => {
        let isCovered = genome.some(sat => angularDistanceDeg(sat.lat, sat.lon, reg.lat, reg.lon) <= 12.0);
        regList.innerHTML += `<div class="list-item ${isCovered ? 'covered' : 'danger'}"><span>${reg.name}</span><span class="badge">${isCovered ? 'CUBIERTO' : 'FALTA'}</span></div>`;
      });
    }

    const fitnessChart = new Chart(document.getElementById('fitnessChart').getContext('2d'), {
      type: 'line',
      data: { labels: [], datasets: [{ data: [], borderColor: '#00e5ff', backgroundColor: 'rgba(0, 229, 255, 0.1)', fill: true, tension: 0.3 }] },
      options: { responsive: true, maintainAspectRatio: false, plugins: { legend: { display: false } }, scales: { x: { display: false }, y: { display: false } } }
    });

    let isEvolving = false, isOrbiting = true;
    async function runGeneticAlgorithm() {
      if(isEvolving) return;
      isEvolving = true;
      let population = Array.from({length: 30}, () => Array.from({length: SAT_COUNT}, () => ({ lat: (Math.random() - 0.5) * 140, lon: (Math.random() - 0.5) * 360, alt: 1.2 })));

      for(let gen = 1; gen <= 60; gen++) {
        population.sort((a, b) => evaluateFitness(b).score - evaluateFitness(a).score);
        currentSats = population[0];
        updateVisuals(currentSats);
        document.getElementById('gen-counter').innerText = `${gen} / 60`;
        fitnessChart.data.labels.push(gen);
        fitnessChart.data.datasets[0].data.push(evaluateFitness(currentSats).score);
        if(gen % 5 === 0) fitnessChart.update();
        await new Promise(r => setTimeout(r, 30));
      }
      isEvolving = false;
    }

    document.getElementById('btn-run-ga').addEventListener('click', runGeneticAlgorithm);
    document.getElementById('btn-reset').addEventListener('click', () => {
      currentSats = Array.from({length: SAT_COUNT}, () => ({ lat: (Math.random() - 0.5) * 140, lon: (Math.random() - 0.5) * 360, alt: 1.2 }));
      updateVisuals(currentSats);
    });
    document.getElementById('btn-toggle-orbit').addEventListener('click', (e) => {
      isOrbiting = !isOrbiting;
      e.target.innerText = isOrbiting ? '⏸ Pausar Órbitas' : '▶ Reanudar Órbitas';
    });

    function animate() {
      requestAnimationFrame(animate);
      if(isOrbiting) earthGroup.rotation.y += 0.0015;
      controls.update();
      renderer.render(scene, camera);
    }

    window.addEventListener('resize', () => {
      camera.aspect = window.innerWidth / window.innerHeight;
      camera.updateProjectionMatrix();
      renderer.setSize(window.innerWidth, window.innerHeight);
    });

    updateVisuals(currentSats);
    animate();
  </script>
</body>
</html>
"""

# Forzar celda a 800 píxeles limpios en Google Colab sin barras recortadas
display(HTML(f'<div style="width: 100%; height: 800px; position: relative;">{html_code}</div>'))
```
